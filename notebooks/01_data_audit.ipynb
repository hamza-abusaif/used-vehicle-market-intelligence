```python
## Condition Score Investigation

The `condition` variable uses an unusual numeric encoding.

Examples include:

- 1 = 1.0
- 2 = 2.0
- 5 = 5.0
- 21 = 2.1
- 35 = 3.5
- 49 = 4.9

Whole-number condition grades appear to be stored as single digits, while decimal condition grades are stored without the decimal point.

This column will need to be normalized during data cleaning. The raw values will remain unchanged.
```

```python
## Extreme Value Investigation

### Odometer

The dataset contains 72 records with an odometer value of exactly 999,999 miles.

Other mileage values above 500,000 occur only once or twice, while 999,999 appears repeatedly. This suggests that 999,999 may represent a special source-system value rather than literal mileage.

The meaning of this value has not been confirmed, so no changes will be made until it is investigated further.

### Selling Price

Several vehicles have recorded selling prices of $1.

Some of these vehicles have MMR values worth thousands or tens of thousands of dollars. These records may represent unusual transactions, data-entry issues, or another special case.

Extremely low prices will be investigated further before they are included in modeling.
```

```python
make_variants = (
    cars.dropna(subset=["make"])
    .assign(make_lower = lambda df: df["make"].str.lower())
    .groupby("make_lower")["make"]
    .agg(lambda x: sorted(set(x)))
)

make_variants[make_variants.apply(len)>1].head(30)
```

```python
body_variants = (
    cars.dropna(subset=["body"])
    .assign(body_lower = lambda df: df["body"].str.lower())
    .groupby("body_lower")["body"]
    .agg(lambda x: sorted(set(x)))
)

body_variants[body_variants.apply(len)>1].head(30)
```

```python
text_columns = ["make", "model", "trim", "body", "transmission", "color", "interior"]

for column in text_columns:
    count = (
        cars[column].notna()
        & (cars[column] != cars[column].str.strip())
    ).sum()

    print(column, count)
```
